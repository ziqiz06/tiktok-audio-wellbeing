# TikTok EMA Project — Ridge Regression Pipeline

## Goal
Test whether session-level behavioral and audio/content features predict EMA outcomes using Ridge regression.

We compare three models of increasing complexity:
- **Model A**: behavioral features only (from DDP logs)
- **Model B**: behavioral + handcrafted audio features (from Librosa, PCA-reduced)
- **Model C**: behavioral + handcrafted audio + AST deep embeddings (768-dim, PCA-reduced)

The unit of analysis is one **EMA burst** (2-hour pre-survey window) per participant.

Ridge regression is appropriate here because:
- We have many raw features relative to observations, even after PCA-reducing the two content blocks
- L2 regularization handles multicollinearity
- Coefficients remain interpretable (on standardized/PCA features — see Section 11)

See `power_analysis.py` for the sample-size reasoning behind PCA-reducing both content blocks — at the original pilot scale, the unreduced 264-dim handcrafted audio block made "does audio help" close to undetectable no matter the sample size.

## 1. Imports and Data Loading

## 4. Feature Grouping

Features are split into three groups based on column name patterns:
- **Behavioral**: DDP-derived event counts and durations
- **Handcrafted audio**: Librosa features aggregated per burst (mean + std)
- **AST embeddings**: 768-dim deep audio representation (mean per burst, PCA-reduced before modeling)

````python
# Columns to always exclude
exclude = {"PID", "Start Date", target, "pilot_n_videos_sampled",
           "pilot_n_videos_with_features"}

all_cols = set(df.columns) - exclude

# Behavioral: DDP-computed burst/person summaries
behav_patterns = ["ddp_", "app_Browser"]
behav_cols = sorted([
    c for c in all_cols
    if any(c.startswith(p) or p in c for p in behav_patterns)
])

# AST: deep embedding dimensions (use only _mean, not _std, for the PCA block)
ast_mean_cols = sorted([c for c in all_cols if c.startswith("ast_") and c.endswith("_mean")])
ast_std_cols  = sorted([c for c in all_cols if c.startswith("ast_") and c.endswith("_std")])
# Use only means for PCA (std adds noise at N=150)
ast_cols = ast_mean_cols

# Handcrafted audio: everything else that looks like an audio feature
audio_patterns = ["mfcc", "chroma", "tempo", "rms_", "zcr_", "tonnetz",
                  "centroid", "rolloff", "bandwidth", "contrast",
                  "beat_count", "non_silence"]
audio_cols = sorted([
    c for c in all_cols
    if any(p in c for p in audio_patterns)
    and not c.startswith("ast_")
])

# Drop any columns that aren't numeric (e.g. ddp_matchtime_* contain timestamp lists)
def keep_numeric(cols, df):
    return [c for c in cols if pd.api.types.is_numeric_dtype(df[c])]

behav_cols = keep_numeric(behav_cols, df)
audio_cols  = keep_numeric(audio_cols, df)
ast_cols    = keep_numeric(ast_cols, df)

print(f"Behavioral features    : {len(behav_cols)}")
print(f"Handcrafted audio      : {len(audio_cols)}")
print(f"AST mean dims (PCA in) : {len(ast_cols)}")
print(f"\nSample behavioral : {behav_cols[:4]}")
print(f"Sample audio      : {audio_cols[:4]}")
print(f"Sample AST        : {ast_cols[:4]}")
````

## 5. Build Three Feature Matrices

- **Model A**: behavior only — establishes a baseline
- **Model B**: behavior + handcrafted audio (PCA-reduced) — tests whether Librosa features add signal
- **Model C**: behavior + handcrafted audio (PCA-reduced) + AST embeddings (PCA-reduced) — tests whether deep representations add signal beyond engineered features

Note: R² estimates at pilot scale can be noisy — check `power_analysis.py` for the sample size needed to trust a given effect size before over-interpreting a comparison between models.

````python
# Filter to columns that actually exist in df
def filter_cols(cols, df):
    present = [c for c in cols if c in df.columns]
    missing = [c for c in cols if c not in df.columns]
    if missing:
        warnings.warn(f"{len(missing)} columns not found and skipped: {missing[:5]}")
    return present

behav_cols = filter_cols(behav_cols, df)
audio_cols  = filter_cols(audio_cols, df)
ast_cols    = filter_cols(ast_cols, df)

# feat_* here are the RAW column sets going into each model's ColumnTransformer;
# the audio and AST blocks get PCA-reduced inside make_pipeline (Section 6),
# so the number of columns below is not the number of features Ridge actually sees.
feat_A = behav_cols
feat_B = behav_cols + audio_cols
feat_C = behav_cols + audio_cols + ast_cols

print(f"Model A raw columns: {len(feat_A)} (all behavioral, no PCA)")
print(f"Model B raw columns: {len(feat_B)} -> behav {len(behav_cols)} + audio {len(audio_cols)} (PCA'd to <= AUDIO_PCA)")
print(f"Model C raw columns: {len(feat_C)} -> behav {len(behav_cols)} + audio {len(audio_cols)} (PCA'd) + ast {len(ast_cols)} (PCA'd to <= N_PCA)")
````

## 6. Preprocessing

- Missing values → mean imputation
- All features → StandardScaler
- Handcrafted audio block → PCA (`AUDIO_PCA` components) applied after scaling
- AST block → PCA (`N_PCA` components) applied after scaling, separately

Both content blocks (audio and AST) get the same PCA treatment for the same reason: with far more raw predictors than observations, an unreduced block makes the model comparison nearly impossible to power adequately (see `power_analysis.py`). Behavioral features are not PCA-reduced — there are few enough of them relative to N that reduction isn't needed.

````python
AUDIO_PCA = 10  # matches power_analysis.py's tested scenario for the audio block
N_PCA = 30      # AST block -- unchanged from the original design
ALPHAS = [0.01, 0.1, 1.0, 10.0, 100.0, 1000.0]

class Winsorizer(BaseEstimator, TransformerMixin):
    """Clips each column to its [lower_pct, upper_pct] percentile range,
    fit on training data only (no leakage across CV folds).

    Added after a real bug, found by running this pipeline at the full
    N=736 scale: one participant's ddp_density_post_person was 1.50 against
    a population range of 0-0.13 (a ~46 SD outlier, likely a near-zero-
    denominator artifact in the raw DDP export). Ridge has no protection
    against linear extrapolation on a held-out point that far outside the
    training distribution -- this one participant's 14 held-out rows got
    predicted around -1050 for a variable bounded [0,100], enough on its
    own to send pooled R² from ~-0.1 (matches the smaller pilot) to ~-29
    (nonsensical). Winsorizing at (5,95) fixed it; (1,99) still left a
    second, less extreme outlier causing the same problem for someone else.
    """
    def __init__(self, lower_pct=5, upper_pct=95):
        self.lower_pct = lower_pct
        self.upper_pct = upper_pct

    def fit(self, X, y=None):
        X = np.asarray(X, dtype=float)
        self.lower_ = np.nanpercentile(X, self.lower_pct, axis=0)
        self.upper_ = np.nanpercentile(X, self.upper_pct, axis=0)
        return self

    def transform(self, X):
        return np.clip(np.asarray(X, dtype=float), self.lower_, self.upper_)

    def get_feature_names_out(self, input_features=None):
        # Winsorizing clips values in place -- doesn't add/drop/reorder
        # columns, so names pass through unchanged. Needed for the
        # coefficient-inspection cell's get_feature_names_out() call.
        if input_features is not None:
            return np.asarray(input_features, dtype=object)
        return np.asarray([f"x{i}" for i in range(len(self.lower_))], dtype=object)

def make_block_pipeline(cols, n_pca, n_samples):
    """impute → winsorize → scale [→ PCA if n_pca is not None].
    PCA n_components is capped at min(n_pca, len(cols), n_samples-1) so it
    never requests more components than rows or raw columns available.
    """
    steps = [
        ("impute",    SimpleImputer(strategy="mean")),
        ("winsorize", Winsorizer()),
        ("scale",     StandardScaler()),
    ]
    if n_pca is not None:
        safe_n = min(n_pca, len(cols), n_samples - 1)
        steps.append(("pca", PCA(n_components=safe_n)))
    return Pipeline(steps)

def make_pipeline(blocks, n_samples):
    """blocks: list of (name, cols, n_pca_or_None). One ColumnTransformer
    branch per block, each independently imputed/winsorized/scaled/
    (optionally PCA'd), then a single RidgeCV over the concatenated result.
    """
    ct = ColumnTransformer([
        (name, make_block_pipeline(cols, n_pca, n_samples), cols)
        for name, cols, n_pca in blocks
    ])
    return Pipeline([
        ("preprocess", ct),
        ("ridge",      RidgeCV(alphas=ALPHAS, scoring="r2")),
    ])

print(f"AUDIO_PCA: {AUDIO_PCA}")
print(f"N_PCA (AST): {N_PCA}")
print(f"Ridge alpha grid: {ALPHAS}")
````

## 7. Cross-Validation

We use **Leave-One-Group-Out (LOGO) CV** grouped by participant ID (`PID`).

- Participants can contribute multiple bursts. LOO by row would let a participant's other bursts into the training fold when one of their bursts is held out — inflating generalization estimates.
- LOGO holds out **all bursts from one participant at a time**, so the model is always evaluated on a person it has never seen during training.
- Number of folds = number of unique participants in the modeling data (printed below) — this is the correct choice if the goal is to generalize to new participants.

````python
from sklearn.model_selection import LeaveOneGroupOut, cross_val_predict
from sklearn.metrics import r2_score, root_mean_squared_error

logo = LeaveOneGroupOut()

# Drop rows where target is missing
df_model = df.dropna(subset=[target]).copy()
y = df_model[target].values
groups = df_model["PID"].values
n = len(df_model)
n_groups = len(np.unique(groups))
print(f"Rows used for modeling : {n}  (dropped {len(df) - n} with missing target)")
print(f"CV strategy            : Leave-One-Group-Out by PID ({n_groups} folds)")

def evaluate_model(pipe, X, y, cv, groups):
    """
    Get out-of-fold predictions via LOGO CV, then compute R² and RMSE
    on the full pooled predictions. This avoids undefined per-fold R²
    (which occurs when a participant contributes only 1 burst to their test fold).
    """
    y_pred = cross_val_predict(pipe, X, y, cv=cv, groups=groups)
    r2   = r2_score(y, y_pred)
    rmse = root_mean_squared_error(y, y_pred)
    residuals = y - y_pred
    return {
        "r2":        r2,
        "rmse":      rmse,
        "y_pred":    y_pred,
        "residuals": residuals,
    }

blocks_A = [("behav", behav_cols, None)]
blocks_B = [("behav", behav_cols, None), ("audio", audio_cols, AUDIO_PCA)]
blocks_C = [("behav", behav_cols, None), ("audio", audio_cols, AUDIO_PCA), ("ast", ast_cols, N_PCA)]

# Model A
pipe_A = make_pipeline(blocks_A, n)
X_A = df_model[feat_A]
res_A = evaluate_model(pipe_A, X_A, y, logo, groups)
print(f"\nModel A — R²: {res_A['r2']:.3f}   RMSE: {res_A['rmse']:.3f}")

# Model B
pipe_B = make_pipeline(blocks_B, n)
X_B = df_model[feat_B]
res_B = evaluate_model(pipe_B, X_B, y, logo, groups)
print(f"Model B — R²: {res_B['r2']:.3f}   RMSE: {res_B['rmse']:.3f}")

# Model C
pipe_C = make_pipeline(blocks_C, n)
X_C = df_model[feat_C]
res_C = evaluate_model(pipe_C, X_C, y, logo, groups)
print(f"Model C — R²: {res_C['r2']:.3f}   RMSE: {res_C['rmse']:.3f}")
````

## 8. Summary Table

````python
def post_pca_count(blocks, n_samples):
    total = 0
    for _, cols, n_pca in blocks:
        total += len(cols) if n_pca is None else min(n_pca, len(cols), n_samples - 1)
    return total

summary = pd.DataFrame({
    "Model": ["A: Behavior", "B: Behavior + Audio(PCA)", "C: Behavior + Audio(PCA) + AST(PCA)"],
    "N features (post-PCA)": [
        post_pca_count(blocks_A, n),
        post_pca_count(blocks_B, n),
        post_pca_count(blocks_C, n),
    ],
    "R² (OOF)":  [res_A['r2'],   res_B['r2'],   res_C['r2']],
    "RMSE (OOF)":[res_A['rmse'], res_B['rmse'], res_C['rmse']],
})
summary = summary.round(4)
display(summary)
````

````python
# PCA explained variance for AST features
X_ast = df_model[ast_cols].copy()
imp = SimpleImputer(strategy="mean")
scl = StandardScaler()
X_ast_imputed = scl.fit_transform(imp.fit_transform(X_ast))

pca_full = PCA(n_components=min(40, len(ast_cols))).fit(X_ast_imputed)
cumvar = np.cumsum(pca_full.explained_variance_ratio_)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(range(1, len(cumvar) + 1), pca_full.explained_variance_ratio_, alpha=0.7, label="Per component")
ax.plot(range(1, len(cumvar) + 1), cumvar, color="red", marker="o", markersize=3, label="Cumulative")
ax.axvline(N_PCA, color="gray", linestyle="--", label=f"n_pca={N_PCA}")
ax.set_title("AST Embedding — PCA Explained Variance")
ax.set_xlabel("Principal Component")
ax.set_ylabel("Explained Variance Ratio")
ax.legend()
plt.tight_layout()
plt.savefig("../results/figures/pca_explained_variance.png", dpi=150)
plt.show()

print(f"Variance explained by first {N_PCA} components: {cumvar[N_PCA-1]:.3f}")
````

## 10. Interpretation

Use this section to reason about what the numbers mean.

### If Model C improves over A and B:
→ Deep audio representations (AST embeddings) carry signal about emotional experience beyond behavioral counts and engineered features. This supports the value of content-level features for EMA prediction.

### If improvements are small or unstable:
→ Check `power_analysis.py` for the power you actually have at the current N before reading much into a small improvement (or lack of one) — an underpowered comparison can't distinguish "no effect" from "effect too subtle to detect at this N."

### If Model A already fits well:
→ Behavioral counts (views, likes, duration) may be sufficient proxies at pilot scale. Audio features may matter more at larger N.

### Red flags to watch:
- R² close to 1 on a small dataset = almost certainly overfitting
- R² < 0 = model performs worse than predicting the mean
- Very large RMSE std = unstable across folds

## 11. Coefficient Inspection (Best Model)

Note: Ridge coefficients are on standardized features, so they reflect relative importance within the model — **not** raw effect sizes.

For Model C, the AST-derived features are PCA components and are **not directly interpretable** in terms of individual audio dimensions.

````python
# Fit Models A and B on full data to inspect coefficients
# (Coefficients on standardized/PCA features — relative importance only, not
# raw effect sizes. For Model B, "audio__pcaN" are PCA components of the
# handcrafted audio block, not individual Librosa features — same caveat
# Model C's AST features already carried, now also true for audio in B.)
for model_label, pipe, cols in [("A", pipe_A, feat_A), ("B", pipe_B, feat_B)]:
    pipe.fit(df_model[cols], y)
    coef = pipe.named_steps["ridge"].coef_
    names = pipe.named_steps["preprocess"].get_feature_names_out()
    coef_df = pd.DataFrame({"feature": names, "coefficient": coef})
    coef_df = coef_df.reindex(coef_df["coefficient"].abs().sort_values(ascending=False).index)

    top_n = 15
    top = coef_df.head(top_n)

    fig, ax = plt.subplots(figsize=(8, 5))
    colors = ["#C44E52" if v < 0 else "#4C72B0" for v in top["coefficient"]]
    ax.barh(top["feature"], top["coefficient"], color=colors)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_title(f"Top {top_n} Ridge Coefficients — Model {model_label} ({target})")
    ax.set_xlabel("Standardized Coefficient (full-data fit)")
    ax.invert_yaxis()
    plt.tight_layout()
    plt.savefig(f"../results/figures/coefficients_model_{model_label}.png", dpi=150)
    plt.show()
````
